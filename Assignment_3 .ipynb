```python
import pandas as pd

# 1. Create a Pandas Series containing numbers from 1 to 10
series = pd.Series(range(1, 11))

# Convert it into a DataFrame with a column named "Numbers"
df = pd.DataFrame({'Numbers': series})

# Print Series
print("Pandas Series:")
print(series)
print("\n" + "-" * 30 + "\n")

# Print DataFrame
print("Pandas DataFrame:")
print(df)
print("\n" + "-" * 30 + "\n")

# Print Shape
print(f"Shape of DataFrame: {df.shape}")

# Print Data type
print(f"Data type of 'Numbers' column: {df['Numbers'].dtype}")
print("\n" + "-" * 30 + "\n")

# Print Sum of elements
print(f"Sum of elements: {df['Numbers'].sum()}")

# Print Mean of elements
print(f"Mean of elements: {df['Numbers'].mean()}")

# Print Maximum and Minimum values
print(f"Maximum value: {df['Numbers'].max()}")
print(f"Minimum value: {df['Numbers'].min()}")
```

##  2. DataFrame Arithmetic

```python
import pandas as pd
import numpy as np

# Create DataFrame A
data_a = {'C1': [1, 4, 7], 'C2': [2, 5, 8], 'C3': [3, 6, 9]}
df_a = pd.DataFrame(data_a, index=['R1', 'R2', 'R3'], columns=['C1', 'C2', 'C3'])

# Create DataFrame B
data_b = {'C1': [9, 6, 3], 'C2': [8, 5, 2], 'C3': [7, 4, 1]}
df_b = pd.DataFrame(data_b, index=['R1', 'R2', 'R3'], columns=['C1', 'C2', 'C3'])

print("DataFrame A:")
print(df_a)
print("\n" + "-" * 30 + "\n")

print("DataFrame B:")
print(df_b)
print("\n" + "-" * 30 + "\n")

# 1. DataFrame Addition
print("DataFrame Addition (A + B):")
print(df_a + df_b)
print("\n" + "-" * 30 + "\n")

# 2. DataFrame Subtraction
print("DataFrame Subtraction (A - B):")
print(df_a - df_b)
print("\n" + "-" * 30 + "\n")

# 3. Element-wise Multiplication
print("Element-wise Multiplication (A * B):")
print(df_a * df_b)
print("\n" + "-" * 30 + "\n")

# 4. Matrix Multiplication (using .dot())
# To perform matrix multiplication, we use the .values attribute to get the underlying NumPy arrays
# and then use numpy's dot product or the @ operator.
print("Matrix Multiplication (A @ B):")
print(pd.DataFrame(df_a.values @ df_b.values, index=df_a.index, columns=df_b.columns))
print("\n" + "-" * 30 + "\n")
```

## 3. Statistics on Student Marks

```python
import pandas as pd

# Create the DataFrame for student marks
data_marks = {
    'Maths': [85, 70, 95, 60, 88],
    'Science': [90, 65, 91, 75, 84],
    'English': [88, 72, 93, 70, 86],
    'History': [92, 68, 96, 80, 90]
}
index_students = ['Amit', 'Bhavna', 'Chirag', 'Divya', 'Esha']
df_marks = pd.DataFrame(data_marks, index=index_students)

print("Original DataFrame of Student Marks:")
print(df_marks)
print("\n" + "-" * 30 + "\n")

# 1. Average marks of each student (row-wise mean)
student_avg_marks = df_marks.mean(axis=1)
print("Average marks of each student:")
print(student_avg_marks)
print("\n" + "-" * 30 + "\n")

# 2. Average marks of each subject (column-wise mean)
subject_avg_marks = df_marks.mean(axis=0)
print("Average marks of each subject:")
print(subject_avg_marks)
print("\n" + "-" * 30 + "\n")

# 3. Highest and lowest marks
highest_mark = df_marks.max().max() # max of all subjects' max
lowest_mark = df_marks.min().min() # min of all subjects' min

print(f"Highest mark: {highest_mark}")
print(f"Lowest mark: {lowest_mark}")
print("\n" + "-" * 30 + "\n")

# 4. Student having the highest total marks
df_marks['Total_Marks'] = df_marks.sum(axis=1)
highest_total_student = df_marks['Total_Marks'].idxmax()
highest_total_score = df_marks['Total_Marks'].max()

print(f"Student with the highest total marks: {highest_total_student} (Total: {highest_total_score})")
print("\n" + "-" * 30 + "\n")

# 5. Sort the DataFrame by total marks in descending order
df_marks_sorted = df_marks.sort_values(by='Total_Marks', ascending=False)
print("DataFrame sorted by Total Marks (descending order):")
print(df_marks_sorted)
print("\n" + "-" * 30 + "\n")
```

## 4.Handling Missing Data using Pandas

```python
import pandas as pd
import numpy as np

# Create the DataFrame for temperature data
data_temp = {
    'Day': ['Day1', 'Day2', 'Day3', 'Day4', 'Day5', 'Day6', 'Day7'],
    'Temperature': [30, np.nan, 32, 29, np.nan, 35, 31]
}
df_temp = pd.DataFrame(data_temp)

print("Original DataFrame with Missing Values:")
print(df_temp)
print("\n" + "-" * 30 + "\n")

# 1. Identify missing values using isnull()
print("Missing values using isnull():")
print(df_temp.isnull())
print("Total missing values: ", df_temp.isnull().sum().sum())
print("\n" + "-" * 30 + "\n")

# Calculate mean temperature for filling
mean_temperature = df_temp['Temperature'].mean()
print(f"Mean Temperature: {mean_temperature:.2f}°C")

# 2. Fill missing values with the mean temperature
df_temp_mean_filled = df_temp.copy()
df_temp_mean_filled['Temperature'] = df_temp_mean_filled['Temperature'].fillna(mean_temperature)
print("DataFrame after filling missing values with mean:")
print(df_temp_mean_filled)
print("\n" + "-" * 30 + "\n")

# 3. Fill missing values using forward fill (ffill)
df_temp_ffill = df_temp.copy()
df_temp_ffill['Temperature'] = df_temp_ffill['Temperature'].ffill()
print("DataFrame after filling missing values with forward fill (ffill):")
print(df_temp_ffill)
print("\n" + "-" * 30 + "\n")

# 4. Drop rows/columns containing missing values
# Dropping rows with any missing values
df_temp_dropped_rows = df_temp.dropna(how='any')
print("DataFrame after dropping rows with missing values:")
print(df_temp_dropped_rows)
print("\n" + "-" * 30 + "\n")

# Dropping columns with any missing values (in this case, it would drop 'Temperature')
df_temp_dropped_cols = df_temp.dropna(axis=1, how='any')
print("DataFrame after dropping columns with missing values:")
print(df_temp_dropped_cols)
print("\n" + "-" * 30 + "\n")

# 5. Calculate the average temperature after cleaning the data
# Let's consider cleaning by dropping missing rows for this calculation
average_temp_after_cleaning = df_temp_dropped_rows['Temperature'].mean()
print(f"Average temperature after cleaning (dropping rows): {average_temp_after_cleaning:.2f}°C")
print("\n" + "-" * 30 + "\n")
```

## 5. Data Analysis using Pandas

```python
import pandas as pd

# Create the DataFrame for sales data
data_sales = {
    'Jan': [120, 100, 150, 90],
    'Feb': [130, 110, 145, 100],
    'Mar': [140, 120, 160, 110],
    'Apr': [150, 130, 170, 120],
    'May': [160, 140, 180, 130],
    'Jun': [170, 150, 190, 140]
}
index_products = ['Product A', 'Product B', 'Product C', 'Product D']
df_sales = pd.DataFrame(data_sales, index=index_products)

print("Original Sales DataFrame:")
print(df_sales)
print("\n" + "-" * 30 + "\n")

# 1. Total sales of each product
# Using sum(axis=1) for row-wise sum
total_sales_per_product = df_sales.sum(axis=1)
print("Total sales of each product:")
print(total_sales_per_product)
print("\n" + "-" * 30 + "\n")

# 2. Average monthly sales of each product
average_monthly_sales_per_product = df_sales.mean(axis=1)
print("Average monthly sales of each product:")
print(average_monthly_sales_per_product)
print("\n" + "-" * 30 + "\n")

# 3. Best-selling product overall
best_selling_product = total_sales_per_product.idxmax()
print(f"Best-selling product overall: {best_selling_product} (Total Sales: {total_sales_per_product.max()})")
print("\n" + "-" * 30 + "\n")

# 4. Month with the highest overall sales
# Sum sales for each month (column-wise sum)
monthly_total_sales = df_sales.sum(axis=0)
highest_sales_month = monthly_total_sales.idxmax()
print(f"Month with the highest overall sales: {highest_sales_month} (Total Sales: {monthly_total_sales.max()})")
print("\n" + "-" * 30 + "\n")

# 5. Standard deviation of sales for each product
std_dev_sales_per_product = df_sales.std(axis=1)
print("Standard deviation of sales for each product:")
print(std_dev_sales_per_product)
print("\n" + "-" * 30 + "\n")

# 6. Create a pivot table showing total sales per product per quarter
# First, melt the DataFrame to long format to easily add a 'Quarter' column
df_melted = df_sales.reset_index().melt(id_vars='index', var_name='Month', value_name='Sales')
df_melted = df_melted.rename(columns={'index': 'Product'})

# Map months to quarters
quarter_map = {'Jan': 'Q1', 'Feb': 'Q1', 'Mar': 'Q1', 'Apr': 'Q2', 'May': 'Q2', 'Jun': 'Q2'}
df_melted['Quarter'] = df_melted['Month'].map(quarter_map)

# Create pivot table
pivot_table_quarterly_sales = df_melted.pivot_table(index='Product', columns='Quarter', values='Sales', aggfunc='sum')
print("Pivot table showing total sales per product per quarter:")
print(pivot_table_quarterly_sales)
print("\n" + "-" * 30 + "\n")
```
