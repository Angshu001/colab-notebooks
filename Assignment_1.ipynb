```python
numbers = [12, 45, 7, 23, 89, 34, 56, 19, 3, 67]
largest = max(numbers)
smallest = min(numbers)
total = sum(numbers)
average = total / len(numbers)
print("List:", numbers)
print("Largest element:", largest)
print("Smallest element:", smallest)
print("Sum of elements:", total)
print("Average of elements:", average)
```

```python
numbers = [10, 20, 30, 40, 50]
print("Original list:", numbers)
position = 2
new_element = 25
numbers.insert(position, new_element)
print("After insertion:", numbers)
numbers.remove(40)
print("After deleting 40:", numbers)
numbers.reverse()
print("Reversed list:", numbers)
```

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print("Second row:", matrix[1])
print("Third column:")
for row in matrix:
    print(row[2])
total = 0
for row in matrix:
    for element in row:
        total += element

print("Sum of all elements:", total)
```

```python
marks = [
    [85, 90, 88],   # Student 1
    [78, 82, 80],   # Student 2
    [92, 95, 94],   # Student 3
    [70, 75, 72]    # Student 4
]
highest_avg = 0
top_student = 0
for i in range(len(marks)):
    avg = sum(marks[i]) / len(marks[i])
    print("Student", i + 1, "Average:", avg)

    if avg > highest_avg:
        highest_avg = avg
        top_student = i + 1
print("\nStudent with the highest average: Student", top_student)
print("Highest Average:", highest_avg)
```

```python
marks = {
    "Math": 95,
    "Science": 88,
    "English": 91,
    "History": 76,
    "Computer": 99
}
print("Subjects:")
print(marks.keys())
print("\nMarks:")
print(marks.values())
highest_subject = max(marks, key=marks.get)
print("\nSubject with the highest marks:")
print(highest_subject, "-", marks[highest_subject])
```

```python
employees = {
    "HR": ["Alice", "Bob"],
    "IT": ["John", "David", "Emma"],
    "Sales": ["Tom", "Jerry"]
}

department = "IT"

print("Employees in", department, ":", employees[department])

print("\nEmployee Count:")
for dept in employees:
    print(dept, ":", len(employees[dept]))
```

```python
students = [
    {"Name": "Amit", "Marks": 85},
    {"Name": "Riya", "Marks": 92},
    {"Name": "Neha", "Marks": 88}
]

print("Student Names:")
for student in students:
    print(student["Name"])

topper = max(students, key=lambda x: x["Marks"])

print("Topper:", topper["Name"], "-", topper["Marks"])
```

```python
books = [
    {"Title": "Python", "Price": 450},
    {"Title": "Java", "Price": 600},
    {"Title": "C++", "Price": 500}
]

expensive = max(books, key=lambda x: x["Price"])

print("Most Expensive Book:", expensive["Title"], "-", expensive["Price"])

total = sum(book["Price"] for book in books)
print("Total Cost:", total)
```
